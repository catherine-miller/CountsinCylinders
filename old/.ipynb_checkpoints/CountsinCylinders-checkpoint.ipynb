```python
#select LRGs and ELGs based on a bitmask
#include a redshift cut: 0.75 < z < 1
zmin = 0.75
zmax = 1
lrgsecondaries = sp[(sp["SV3_DESI_TARGET"]&0x1==0x1)&(sp["ZWARN"]==0)&(sp["Z"] > 0.75)&(sp["Z"] < 1)]
elgsecondaries = sp[(sp["SV3_DESI_TARGET"]&0x2==0x2)&(sp["ZWARN"]==0)&(sp["Z"] > 0.75)&(sp["Z"] < 1)]
both = sp[(sp["SV3_DESI_TARGET"]&0x3==0x3)&(sp["ZWARN"]==0)&(sp["Z"] > 0.75)&(sp["Z"] < 1)]
print("Number of LRGs "+str(zmin)+" < z < "+str(zmax)+": "+str(len(lrg)))
print("Number of ELGs "+str(zmin)+" < z < "+str(zmax)+": "+str(len(elg)))
print("Number of galaxies identified as both ELG and LRG: "+str(len(both)))
```

```python
def raDecToUnitSphere(a, d):
    a = a*np.pi/180
    d = d*np.pi/180
    x = np.cos(a)*np.cos(d)
    y = np.sin(a)*np.cos(d)
    z = np.sin(d)
    return x,y,z

def raDecToCartesian(a, d, r):
    return r*raDecToUnitSphere(a,d)
```

```python
#get comoving distances of galaxies along line of sight
elgsecondaries['d'] = cosmo.comoving_distance(elgsecondaries['Z'])
lrgsecondaries['d'] = cosmo.comoving_distance(lrgsecondaries['Z'])

'''
#get coordinates of galaxies on the unit sphere
elg['x_us'],elg['y_us'],elg['z_us'] = raDecToUnitSphere(elg['TARGET_RA'],elg['TARGET_DEC'])
lrg['x_us'],lrg['y_us'],lrg['z_us'] = raDecToUnitSphere(lrg['TARGET_RA'],lrg['TARGET_DEC'])
'''

#get Cartesian coordinates of galaxies
elgsecondaries['x'],elgsecondaries['y'],elgsecondaries['z'] = raDecToCartesian(elg['TARGET_RA'],elg['TARGET_DEC'],elg['d'])
lrgsecondaries['x'],lrgsecondaries['y'],lrgsecondaries['z'] = raDecToCartesian(lrg['TARGET_RA'],lrg['TARGET_DEC'],lrg['d'])
```

```python
### Cylinder Shape ###
'''
Time to define our cylinders. These are actually truncated cones, centered on each galaxy, defined by a maximum angular separation $\theta_{CiC}$ and a maximum line-of-sight separation $L_{CiC}$. For the moment I will just follow a method from $\href{https://arxiv.org/pdf/2309.08675}{\text{this CiC paper}}$, defining $\theta_{CiC}$ so that the volume of each cylinder is $2\pi R^2_{CiC}L_{CiC}$:

$$\theta_{CiC}=\text{arccos}\left(1-\frac{3R^2_{CiC}L_{CiC}}{(d+L_{CiC})^3-(d-L_{CiC})^3}\right)$$

The parameters of each cylinder, its maximum angular separation and minimum and maximum line-of-sight distance, obviously differ for each galaxy.

We can find the angular separation between two galaxies from their right ascension $\phi$ and declination $\theta$:

$$\psi = \arccos(\sin\theta_1+\sin\theta_2+\cos\theta_1\cos\theta_2\cos(\phi_1-\phi_2))$$

Combining these we make an angular separation requirement:

$$\psi < \theta_\text{CiC} $$
$$\to \sin\theta_1+\sin\theta_2+\cos\theta_1\cos\theta_2\cos(\phi_1-\phi_2) > 1-\frac{3R^2_{CiC}L_{CiC}}{(d+L_{CiC})^3-(d-L_{CiC})^3} $$
'''
### Search Method ###
We will search for galaxy pairs with cKDTree's query_ball_point. First we will convert right ascension, declination, and redshift to Cartesian coordinates for each galaxy. Then we will find all the other galaxies in a $\sqrt{(L_{CiC}/2)^2+R_{CiC}}$ Mpc/h sphere around that galaxy. (Ah! come to think of it, I have clipped the edge of my proposed truncated cone...) Next we will prune out neighbors that are not in the truncated cone defined by $\theta_{CiC}$ and $L_{CiC}$.
```

### Cylinders ###
For each galaxy, we define a (true) cylinder with axis along the line of sight of radius $R_{CiC}$ and height $L_{CiC}$. For this first analysis we choose $R_{CiC} = 1 \text{ Mpc/h}$, $L_{CiC} = 20 \text{ Mpc/h}$.

### Search Method ###
We will search for galaxy pairs with cKDTree's query_ball_point. First we will convert right ascension, declination, and redshift to Cartesian coordinates for each galaxy. Then we will find all the other galaxies in a sphere of radius $\sqrt{(L_{CiC}/2)^2+R_{CiC}}$ around that galaxy. Next we will prune out neighbor galaxies not in the truncated cone.

```python
def checkAngularSep(phi1,phi2,theta1,theta2,psimin,psimax):
    #arguments: RA1, RA2, Dec1, Dec2, min separation, max separation (in DEGREES)
    theta1 = theta1*np.pi/180
    theta2 = theta2*np.pi/180
    phi1 = phi1*np.pi/180
    phi2 = phi2*np.pi/180
    psimin = psimin*np.pi/180
    psimax = psimax*np.pi/180
    cospsi = np.sin(theta1)*np.sin(theta2)+np.cos(theta1)*np.cos(theta2)*np.cos(phi1-phi2)
    return (cospsi**2 > np.cos(psimax)**2)&(cospsi**2 < np.cos(psimin)**2)
def rosetteOR(phi, theta, psimin, psimax, rosettes):
    return np.any([checkAngularSep(rosettes['RA'][i],phi,rosettes['Dec'][i],theta,psimin,psimax)
                   for i in range(len(rosettes['RA']))],axis=0)
```

```python
rosettes = {'RA': [150.100, 179.600, 183.100, 189.900, 194.750, 210.000, 215.500, 217.800, 216.300, 219.800, 218.050, 
             242.750, 241.050, 245.880, 252.500, 269.730, 194.750, 212.800, 269.730, 236.100],
            'Dec': [2.182, 0.000, 0.000, 61.800, 28.200, 5.000, 52.500, 34.400, 0.600, 0.600, 2.430, 54.980, 43.450, 43.450,
              34.500, 66.020, 24.700, 0.600, 62.520, 43.450] }
```

```python
psimin = 0.2
psimax = 1.5
elgprimaries = elg[rosetteOR(elg['TARGET_RA'],elg['TARGET_DEC'],psimin,psimax,rosettes)]
lrgprimaries = lrg[rosetteOR(lrg['TARGET_RA'],lrg['TARGET_DEC'],psimin,psimax,rosettes)]
```

```python
#try plotting some rosettes just to make sure we have what we want
plt.figure()
plt.xlim(rosettes['RA'][1]-2,rosettes['RA'][1]+2)
plt.ylim(rosettes['Dec'][1]-2,rosettes['Dec'][1]+2)
plt.title(r"Rosette Radial Cut (ELGs) in SV3: $0.2 < \Theta < 1.5$ ")
plt.scatter(elg["TARGET_RA"],elg["TARGET_DEC"],s=1,label="Excluded from Primaries (Eligible Neighbors)")
plt.scatter(elgprimaries["TARGET_RA"],elgprimaries["TARGET_DEC"],s=1,label="Primaries")
plt.legend()
plt.xlabel("Right Ascension (Degrees)")
plt.ylabel("Declination (Degrees)")

print(len(elgprimaries))
```

```python
#make 3D maps of galaxies
#primaries
elgprimarymap = np.transpose([elgprimaries['x'],elgprimaries['y'],elgprimaries['z']])
lrgprimarymap = np.transpose([lrgprimaries['x'],lrgprimaries['y'],lrgprimaries['z']])
#secondaries
elgsecondarymap = np.transpose([elg['x'],elg['y'],elg['z']])
lrgsecondarymap = np.transpose([lrg['x'],lrg['y'],lrg['z']])
#make ckdtrees
elgtree = cKDTree(elgsecondarymap)
lrgtree = cKDTree(lrgsecondarymap)
```

```python
cylh = 20*cosmo.h #cylinder height in Mpc
cyld = 1*cosmo.h #cylinder diameter in Mpc
rsphere = np.sqrt((cylh/2)**2+(cyld/2)**2) #radius of the sphere of our initial search
```

```python
#search for neighbors
elgprimaries['neighbors'] = elgtree.query_ball_point(elgprimarymap, rsphere)
lrgprimaries['neighbors'] = lrgtree.query_ball_point(lrgprimarymap, rsphere)

#search for neighbors among other class of galaxies
elgprimaries['lrg_neighbors'] = lrgtree.query_ball_point(elgprimarymap, rsphere)
lrgprimaries['elg_neighbors'] = elgtree.query_ball_point(lrgprimarymap, rsphere)
```

```python
#define pruning algorithm
hsq = cylh**2
dsq = cyld**2

def checkcylinder(x1,y1,z1,d1,x2,y2,z2,d2):
    #check if two points ar in cylinder with axis along line of sight
    dDsq = (d1 - d2)**2
    if dDsq > hsq: return False
    elif (x1 - x2)**2 + (y1 - y2)**2 + (z1 - z2)**2 - dDsq > dsq: 
        return False
    else: return True

def tableprune(table1, table2, i, j):
    #i is the index of the central galaxy (in table 1)
    #j is the index of its neighbors (in table 2)
    return checkcylinder(table1[i]['x'],table1[i]['y'],table1[i]['z'],table1[i]['d'],
                         table2[j]['x'],table2[j]['y'],table2[j]['z'],table2[j]['d'])
```

```python
#prune trees:
elg['neighborspruned'] = [np.array(elg[i]['neighbors'])[[tableprune(elg,elg,i,j) for j in elg[i]['neighbors']]][1:] for i in range(len(elg))]
lrg['neighborspruned'] = [np.array(lrg[i]['neighbors'])[[tableprune(lrg,lrg,i,j) for j in lrg[i]['neighbors']]][1:] for i in range(len(lrg))]
elg['lrg_neighborspruned'] = [np.array(elg[i]['lrg_neighbors'])[[tableprune(elg,lrg,i,j) for j in elg[i]['lrg_neighbors']]] for i in range(len(elg))]
lrg['elg_neighborspruned'] = [np.array(lrg[i]['elg_neighbors'])[[tableprune(lrg,elg,i,j) for j in lrg[i]['elg_neighbors']]] for i in range(len(lrg))]
```

```python
#count up the galaxies in cylinders
elg['N_CiC'] = [len(n) for n in elg['neighborspruned']]
lrg['N_CiC'] = [len(n) for n in lrg['neighborspruned']]
elg['N_lrgCiC'] = [len(n) for n in elg['lrg_neighborspruned']]
lrg['N_elgCiC'] = [len(n) for n in lrg['elg_neighborspruned']]
```

```python
bins = np.arange(zmin,zmax,0.05)
elgNCiCavg = [np.avg(elg[(elg['Z'] > bins[b])&(elg['Z'] < bins[b+1])]['N_CiC']) for b in range(len(bins)-1)]

plt.figure()
plt.plot(bins,elgNCiCavg) #should these be uncorrelated?
plt.title("ELG Counts in Cylinders vs. Redshift")
plt.ylabel(r'N_{CiC}')
plt.xlabel('z')
plt.yscale('log')
'''
plt.figure()
plt.scatter(lrg['Z'],lrg['N_CiC']) #should these be uncorrelated?
plt.title("LRG Counts in Cylinders vs. Redshift")
plt.ylabel(r'N_{CiC}')
plt.xlabel('z')
plt.yscale('log')

plt.figure()
plt.title("ELG and LRG Redshifts in DESI SV3")
plt.hist(elg["Z"],bins=100,label="ELG",alpha=0.7)
plt.xlabel("Redshift")
plt.ylabel("Galaxies")

plt.hist(lrg["Z"],bins=100,label="LRG",alpha=0.7)
plt.legend()
plt.yscale("log")'''
```
