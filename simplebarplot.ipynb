```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib

matplotlib.rcParams['mathtext.fontset'] = 'stix'
matplotlib.rcParams['font.family'] = 'STIXGeneral'
matplotlib.fontsize=20
```

```python
data_SV3 = [0.094,0.261,0.102,0.054]
data_Abacus = [0.104,0.259,0.098,0.053]
error_SV3 = [0.002,0.010,0.004,0.002]
error_Abacus = [0.0002,0.001,0.0001,0.0002]
plt.figure()
categories = ["ELG counts \naround ELGs","LRG counts \naround LRGs","ELG counts \naround LRGs","LRG counts \naround ELGs"]
#plt.xlabel(categories)
plt.bar(categories,data_SV3,yerr=error_SV3,edgecolor='blue',capsize=2)
plt.bar(categories,data_Abacus,yerr=error_Abacus,hatch='///',color='none',edgecolor='black',capsize=2)
plt.ylabel("Average Counts",fontsize=15)
plt.title("Average Counts in Cylinders in SV3 and Mock Catalogs")
plt.savefig("old/barplot.png")
```
